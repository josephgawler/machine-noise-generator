```python
import os
import glob
import librosa
import numpy as np
import matplotlib.pyplot as plt

# 配置参数
SR = 16000            # 采样率
AUDIO_DURATION = 5    # 固定音频时长为 5 秒
N_FFT = 1024
HOP_LENGTH = 512
N_MELS = 128

def generate_png_from_wav(src_dir="data/trimmed_fan", dest_dir="data/trimmed_fan_png"):

    print(f"Starting to process wav files in {src_dir}")

    for label in ["normal", "abnormal"]:
        dest_label_dir = os.path.join(dest_dir, label)
        os.makedirs(dest_label_dir, exist_ok=True)
        print(f"Created directory: {dest_label_dir}")

    file_counter = 0
    for label in ["normal", "abnormal"]:
        src_label_dir = os.path.join(src_dir, label)
        if not os.path.exists(src_label_dir):
            print(f"Warning: Directory not found: {src_label_dir}")
            continue
        print(f"Processing label '{label}' in {src_label_dir}")
        
        for root, dirs, files in os.walk(src_label_dir):
            print(f"Entering directory: {root}, Files found: {len(files)}")
            for file in files:
                
                if file.lower().endswith(".wav"):
                    file_path = os.path.join(root, file)
                    print(f"Processing {file_path}...")
                    try:
                        
                        audio, sr = librosa.load(file_path, sr=SR, duration=AUDIO_DURATION)
                        target_length = SR * AUDIO_DURATION
                        if len(audio) < target_length:
                            audio = np.pad(audio, (0, target_length - len(audio)), mode='constant')
                        elif len(audio) > target_length:
                            audio = audio[:target_length]
                        
                        
                        mel_spec = librosa.feature.melspectrogram(
                            y=audio,
                            sr=SR,
                            n_fft=N_FFT,
                            hop_length=HOP_LENGTH,
                            n_mels=N_MELS
                        )
                        
                        log_mel_spec = librosa.power_to_db(mel_spec)
                        
                        norm_spec = (log_mel_spec - log_mel_spec.min()) / (log_mel_spec.max() - log_mel_spec.min())
                        
                        plt.figure(figsize=(10, 4))
                        plt.imshow(norm_spec, origin='lower', aspect='auto', cmap='viridis')
                        plt.axis('off')
                        
                        base_name = os.path.splitext(file)[0]
                        dest_label_dir = os.path.join(dest_dir, label)
                        dest_path = os.path.join(dest_label_dir, base_name + ".png")
                        plt.savefig(dest_path, bbox_inches="tight", pad_inches=0)
                        plt.close()
                        
                        print(f"Saved {dest_path}")
                        file_counter += 1
                    except Exception as e:
                        print(f"Error processing {file_path}: {e}")
    print(f"Processing complete. Total files processed: {file_counter}")

if __name__ == '__main__':
    generate_png_from_wav("data/trimmed_fan", "data/trimmed_fan_png")
```
